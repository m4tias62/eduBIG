## 1. Importar librerías

```python
import pandas as pd

denuncias_24 = pd.read_csv('../raw/20241231_DENUNCIAS_2024_20260331_PUBL.csv', sep=';', encoding='latin-1')

denuncias_25 = pd.read_csv('../raw/20251231_DENUNCIAS_2025_20260331_PUBL.csv', sep=';', encoding='latin-1')

idps_2m_25 = pd.read_csv('../raw/idps2M2025_rbd_preliminar.csv', sep=';', encoding='latin-1')

idps_4b_25 = pd.read_csv('../raw/idps4B2025_rbd_preliminar.csv', sep=';', encoding='latin-1')

idps_8b_25 = pd.read_csv('../raw/idps8B2025_rbd_preliminar.csv', sep=';', encoding='latin-1')

simce_2m_25 = pd.read_csv('../raw/simce2m2025_rbd_preliminar.csv', sep=';', encoding='latin-1')

simce_4b_25 = pd.read_csv('../raw/simce4b2025_rbd_preliminar.csv', sep=';', encoding='latin-1')

simce_6b_24 = pd.read_csv('../raw/simce6b2024_rbd_final.csv', sep=';', encoding='latin-1')

simce_8b_25 = pd.read_csv('../raw/simce8b2025_rbd_preliminar.csv', sep=';', encoding='latin-1')

colegios_base = pd.read_json('../../web-app/public/data/colegios_base.json')
```

```python
print('denuncias_24: ', denuncias_24.shape)
print('denuncias_25: ', denuncias_25.shape)
print('idps_2m_25: ', idps_2m_25.shape)
print('idps_4b_25: ', idps_4b_25.shape)
print('idps_8b_25: ', idps_8b_25.shape)
print('simce_2m_25: ', simce_2m_25.shape)
print('simce_4b_25: ', simce_4b_25.shape)
print('simce_6b_24: ', simce_6b_24.shape)
print('simce_8b_25: ', simce_8b_25.shape)
print('colegios_base: ', colegios_base.shape)
```

```python
denuncias_25_limpia = denuncias_25[['agno', 'rbd']]

denuncias_24_limpia = denuncias_24[['agno', 'rbd']]
```

```python
denuncias_24_limpia = denuncias_24_limpia[denuncias_24_limpia['rbd'].str.strip() != '']

denuncias_24_limpia['rbd'] = denuncias_24_limpia['rbd'].str.strip().astype('int64')

denuncias_25_limpia = denuncias_25_limpia[denuncias_25_limpia['rbd'].str.strip() != '']

denuncias_25_limpia['rbd'] = denuncias_25_limpia['rbd'].str.strip().astype('int64')
```

```python
denuncias_agrupadas_25 = denuncias_25_limpia.groupby('rbd').size().reset_index(name='conteo_denuncias_25')

denuncias_agrupadas_24 = denuncias_24_limpia.groupby('rbd').size().reset_index(name='conteo_denuncias_24')
```

```python
print(denuncias_agrupadas_24.shape)
print(denuncias_agrupadas_25.shape)
```

2.1 Limpieza de datos / idps

```python
idps_2m_25.head()
```

```python
idps_2m_25_limpia = idps_2m_25[['rbd', 'id_indicador', 'prom', 'dif', 'difgru', 'sigdifgru']]

idps_4b_25_limpia = idps_4b_25[['rbd', 'id_indicador', 'prom', 'dif', 'difgru', 'sigdifgru']]

idps_8b_25_limpia = idps_8b_25[['rbd', 'id_indicador', 'prom', 'dif', 'difgru', 'sigdifgru']]
```

```python
idps_2m_25_pivot = idps_2m_25_limpia.pivot_table(index='rbd', columns='id_indicador', values=['prom', 'dif', 'difgru', 'sigdifgru'], aggfunc= 'mean').reset_index()

idps_2m_25_pivot.head()
```

```python
idps_2m_25_pivot.columns =['rbd', 
'autoestima_dif_2m', 'clima_dif_2m', 'participacion_dif_2m', 'habitos_dif_2m', 
'autoestima_difgru_2m', 'clima_difgru_2m', 'participacion_difgru_2m', 'habitos_difgru_2m', 
'autoestima_prom_2m', 'clima_prom_2m', 'participacion_prom_2m', 'habitos_prom_2m', 
'autoestima_sigdifgru_2m', 'clima_sigdifgru_2m', 'participacion_sigdifgru_2m', 'habitos_sigdifgru_2m'
]

idps_2m_25_pivot.shape
```

```python
idps_4b_25_pivot = idps_4b_25_limpia.pivot_table(index='rbd', columns='id_indicador', values=['prom', 'dif', 'difgru', 'sigdifgru'], aggfunc= 'mean').reset_index()

idps_8b_25_pivot = idps_8b_25_limpia.pivot_table(index='rbd', columns='id_indicador', values=['prom', 'dif', 'difgru', 'sigdifgru'], aggfunc= 'mean').reset_index()
```

```python
idps_4b_25_pivot.columns = ['rbd', 
'autoestima_dif_4b', 'clima_dif_4b', 'participacion_dif_4b', 'habitos_dif_4b', 
'autoestima_difgru_4b', 'clima_difgru_4b', 'participacion_difgru_4b', 'habitos_difgru_4b', 
'autoestima_prom_4b', 'clima_prom_4b', 'participacion_prom_4b', 'habitos_prom_4b', 
'autoestima_sigdifgru_4b', 'clima_sigdifgru_4b', 'participacion_sigdifgru_4b', 'habitos_sigdifgru_4b'
]

idps_8b_25_pivot.columns = ['rbd', 
'autoestima_dif_8b', 'clima_dif_8b', 'participacion_dif_8b', 'habitos_dif_8b', 
'autoestima_difgru_8b', 'clima_difgru_8b', 'participacion_difgru_8b', 'habitos_difgru_8b', 
'autoestima_prom_8b', 'clima_prom_8b', 'participacion_prom_8b', 'habitos_prom_8b', 
'autoestima_sigdifgru_8b', 'clima_sigdifgru_8b', 'participacion_sigdifgru_8b', 'habitos_sigdifgru_8b'
]

print(idps_8b_25_pivot.shape)
print(idps_4b_25_pivot.shape)
```

2.2 Limpieza de datos / simce

```python
simce_2m_25_limpia = simce_2m_25[['rbd', 'cod_grupo', 'prom_lect2m_rbd', 'prom_mate2m_rbd',
                                    'difgru_lect2m_rbd', 'difgru_mate2m_rbd',
                                    'siggru_lect2m_rbd', 'siggru_mate2m_rbd']]

simce_4b_25_limpia = simce_4b_25[['rbd', 'cod_grupo', 'prom_lect4b_rbd', 'prom_mate4b_rbd',
                                    'difgru_lect4b_rbd', 'difgru_mate4b_rbd',
                                    'siggru_lect4b_rbd', 'siggru_mate4b_rbd']]

simce_6b_24_limpia = simce_6b_24[['rbd', 'cod_grupo', 'prom_lect6b_rbd', 'prom_mate6b_rbd',
                                    'difgru_lect6b_rbd', 'difgru_mate6b_rbd',
                                    'siggru_lect6b_rbd', 'siggru_mate6b_rbd']]

simce_8b_25_limpia = simce_8b_25[['rbd', 'cod_grupo', 'prom_lect8b_rbd', 'prom_mate8b_rbd', 'prom_hist8b_rbd',
                                    'difgru_lect8b_rbd', 'difgru_mate8b_rbd', 'difgru_hist8b_rbd',
                                    'siggru_lect8b_rbd', 'siggru_mate8b_rbd', 'siggru_hist8b_rbd']]
```

```python
simce_2m_25_limpia = simce_2m_25_limpia.rename(columns={'cod_grupo': 'cod_grupo_2m'})

simce_4b_25_limpia = simce_4b_25_limpia.rename(columns={'cod_grupo': 'cod_grupo_4b'})

simce_6b_24_limpia = simce_6b_24_limpia.rename(columns={'cod_grupo': 'cod_grupo_6b'})

simce_8b_25_limpia = simce_8b_25_limpia.rename(columns={'cod_grupo': 'cod_grupo_8b'})
```

2.3 Limpieza de datos / colegios_base

```python
colegios_base.columns.tolist()
```

```python
colegios_base_limpio = colegios_base.rename(columns={'RBD': 'rbd'})
```

```python
colegios_base_limpio.columns.tolist()
```

```python
colegios_base_limpio['LATITUD'] = colegios_base_limpio['LATITUD'].str.replace(',', '.')
colegios_base_limpio['LATITUD'] = pd.to_numeric(colegios_base_limpio['LATITUD'], errors='coerce')

colegios_base_limpio['LONGITUD'] = colegios_base_limpio['LONGITUD'].str.replace(',', '.')
colegios_base_limpio['LONGITUD'] = pd.to_numeric(colegios_base_limpio['LONGITUD'], errors='coerce')
```

```python
print(colegios_base_limpio['LATITUD'])
print(colegios_base_limpio['LONGITUD'])
```

3. Merge de los datos (utilizando 'rbd' como unificador)

```python
colegios_enriquecidos = (colegios_base_limpio
.merge(simce_2m_25_limpia, on='rbd', how='left')
.merge(simce_4b_25_limpia, on='rbd', how='left')
.merge(simce_6b_24_limpia, on='rbd', how='left')
.merge(simce_8b_25_limpia, on='rbd', how='left')
.merge(denuncias_agrupadas_24, on='rbd', how='left')
.merge(denuncias_agrupadas_25, on='rbd', how='left')
.merge(idps_2m_25_pivot, on='rbd', how='left')
.merge(idps_4b_25_pivot, on='rbd', how='left')
.merge(idps_8b_25_pivot, on='rbd', how='left')
)
```
