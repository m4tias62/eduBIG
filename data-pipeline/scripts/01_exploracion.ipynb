```python
import pandas as pd
```

```python
ruta_archivo = "raw/directorio_mineduc.csv"
df_colegios = pd.read_csv(ruta_archivo, sep=";")
print(df_colegios.shape)
```

```python
pd.set_option("display.max_columns", None)
df_colegios.head()
```

```python
df_pudahuel = df_colegios[df_colegios["NOM_COM_RBD"] == "PUDAHUEL"]
print(f"Número de colegios encontrados en Pudahuel: {df_pudahuel.shape[0]}")
```

```python
print(df_pudahuel.columns.tolist())
```

```python
columnas_clave = [
'RBD',  
'NOM_RBD',  
'ESTADO_ESTAB', 
'NOM_COM_RBD', 
'ORI_RELIGIOSA', 
'ORI_OTRO_GLOSA', 
'PAGO_MATRICULA',
'PAGO_MENSUAL', 
'COD_DEPE',
'LATITUD', 
'LONGITUD', 
'RURAL_RBD', 
'MAT_TOTAL' 
]

df_edubig_base = df_pudahuel[columnas_clave]
df_edubig_base.head()

```

```python
import os

ruta_salida = "../web-app/public/data"
os.makedirs(ruta_salida, exist_ok=True)

df_edubig_base.to_json(f"{ruta_salida}/colegios_base.json", orient='records', force_ascii=False)
```
